# Palindrome Number

An object is a palindrome when it reads the same forward and backward.

> For example, 121 is a palindrome while 123 is not.

Given an integer `x`, return `true` if `x` is a palindrome, and `false` otherwise.

 
```txt
Example 1:

Input: x = 121
Output: true
Explanation: 121 reads as 121 from left to right and from right to left.
```

```txt
Example 2:

Input: x = -121
Output: false
Explanation: From left to right, it reads -121. From right to left, it becomes 121-. Therefore it is not a palindrome.
```

```txt
Example 3:

Input: x = 10
Output: false
Explanation: Reads 01 from right to left. Therefore it is not a palindrome.
```

```txt
Constraints: -231 <= x <= 231 - 1
```

## Step 2: Compare if both are qual or not

````python
def ReverseNum(num):
    # Storing the original number since with every step original number will be reduced to 0 after every iteration
    not_rev = num 
    rev = 0
    while num > 0:
        digit = num % 10
        rev = (rev * 10) + digit
        num = num // 10
    if rev == not_rev:
        return True
    return False
````

# Actual Implementation

````python
class Solution:
    def isPalindrome(self, number: int) -> bool:
        
        if number < 0:
            return False

        original_number  = number 
        
        reverse = 0  
            
        while number > 0: 
                
            last_digit  = number % 10 
                
            reverse = (reverse * 10) + last_digit
                
            number = number // 10
    
        return reverse == original_number
````
