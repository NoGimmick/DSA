# Matrix Reloaded

## Problem statement

Neo has a complex matrix script. The matrix script is a N x M grid of strings. It consists of alphanumeric characters, spaces and symbols (!,@,#,$,%,&). 

To decode the script, Neo needs to read each column and select only the alphanumeric characters and connect them. Neo reads the column from top to bottom and starts reading from the leftmost column.

If there are symbols or spaces between two alphanumeric characters of the decoded script, then Neo replaces them with a single space '' for better readability.

Neo feels that there is no need to use 'if' conditions for decoding.

Alphanumeric characters consist of: [A-Z, a-z, and 0-9].

#### Input Format

- The first line contains space-separated integers  (rows) and  (columns) respectively.
- The next  lines contain the row elements of the matrix script.

> **NOTE:** NO points will be awarded for using if condition


#### Sample Input 0

```txt
7 3
Tsi
h%x
i #
sM 
$a 
#t%
ir!
```

#### Sample Output 0
```
This is Matrix#  %!
```

#### Explanation 0

**Decoded Script:**
```
This$#is% Matrix#  %!
```

> Here, Neo replaces the symbols or spaces between two alphanumeric characters with a single space   ' ' for better readability. So, the final decoded script is:

This is Matrix#  %!

## Key Points

- Input -> Rows (N) x Column (M)
- Input -> Contains N lines of elements of matrix
- Here, 0 < n,m < 100
- Do not use `if Condition`
- To decode : Replace symbols/spaces b/w two alphanumeric char with `''`

# Part - 1: Decode Matrix

````python
string = ''
for c in range(len(matrix[0])):
    for r in range(len(matrix)):
        string = string + matrix[r][c]
print (string)
````

````python
import re

string = "a$#b%&c"

# string = 'a$#b$#c'

string = re.sub(
    r"(?<=[A-Za-z0-9])[^A-Za-z0-9]+(?=[A-Za-z0-9])",
    " ",
    string
)

print (string)
````

# Overall Code

```python
#!/bin/python3

import re

first_multiple_input = input().rstrip().split()

n = int(first_multiple_input[0])

m = int(first_multiple_input[1])

matrix = []

for _ in range(n):
    matrix_item = input()
    matrix.append(matrix_item)
    
string = ''

for c in range(len(matrix[0])):

    for r in range(len(matrix)):

        string = string + matrix[r][c]
            
string = re.sub(
    r"(?<=[A-Za-z0-9])[^A-Za-z0-9]+(?=[A-Za-z0-9])",
    " ",
    string
)

print (string)
```
