```python
def absDiff(elements: list) -> int:

    maximum = float('-inf') 
    
    n = len(elements)
    
    for i in range (n - 1):
        for j in range (i + 1, n):
            diff = abs(elements[i] - elements[j])

            maximum = max(maximum, diff)
            
    return maximum

elements = [1, 2, 5, 8]

print (absDiff(elements))
```

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None 
class Solution: 
    def display(self,head):
        current = head
        while current:
            print(current.data,end=' ')
            current = current.next

    def insert(self, head, data):
        
        new_node = Node(data)

        if head is None:
            return new_node

        current = head

        while current.next is not None:
            current = current.next
        current.next = new_node
        return head

mylist= Solution()
T=int(input())
head=None
for i in range(T):
    data=int(input())
    head=mylist.insert(head,data)    
mylist.display(head); 	  
```

```python
def merge_the_tools(string, k):
    
    for iteration in range (0, len(string), k):
        
        chunk = string[iteration : iteration + k]
    
        seen = ''
    
        for char in chunk:
            if char not in seen:
                seen += char
        
        print (seen)
        
string = 'AABCAAADA'
k = 3

merge_the_tools(string, k)
```
