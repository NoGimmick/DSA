```python
class Binary():

    def ConBin(self,num: int):

        string = ''
        while num > 0:
            string  += str(num % 2) 
            num = num // 2
        return string[::-1]

    def CountNums(self, num, num_to_count: int): 
        '''
        This function counts series of consecutive 1s or 0s
        Num_to_count is always either 1 or 0
        '''
        if num_to_count in [1,0]:
            binary = self.ConBin(num)
            count = 0
            maximum = 0
            for i in binary:
                if int(i) == num_to_count:
                    count = count + 1
                    maximum = max(count, maximum)
                elif int(i) == 0:
                    count = 0
            return maximum
        else:
            raise ValueError('Value must be 1 or 0')

Problem = Binary()

print (Problem.CountNums(87, 1)) # 1010111
```

```python
class FlamesGame:
    def __init__(self):
        # The full form dictionary we created
        self.meanings = {
            'F': 'Friends',
            'L': 'Love',
            'A': 'Affection',
            'M': 'Marriage',
            'E': 'Enemies',
            'S': 'Siblings / Sister'
        }

    def play(self, name1: str, name2: str) -> str:
        # 1. Clean the names (lowercase and remove spaces)
        n1 = list(name1.lower().replace(" ", ""))
        n2 = list(name2.lower().replace(" ", ""))
        
        # 2. Cancel out common letters
        for char in n1[:]:  # Use a copy slice [:] to safe-iterate while modifying
            if char in n2:
                n1.remove(char)
                n2.remove(char)
                
        # 3. Calculate total remaining count
        count = len(n1) + len(n2)
        
        # Edge case: If no letters are left or no names entered
        if count == 0:
            return "No unique letters found to calculate relationship."

        # 4. Cycle through FLAMES and eliminate letters
        flames_list = list("FLAMES")
        index = 0
        
        while len(flames_list) > 1:
            # Determine the index to eliminate using modulo arithmetic
            index = (index + count - 1) % len(flames_list)
            flames_list.pop(index)
            
        # 5. Get the final resulting letter and map it to its meaning
        final_letter = flames_list[0]
        return self.meanings[final_letter]

# --- Let's run the game ---
game = FlamesGame()

print("🔥 Welcome to the FLAMES Game 🔥\n")
player1 = 'Saksham Trivedi' # input("Enter Name 1: ")
player2 = 'Aarti Kulkarni' # input("Enter Name 2: ")

result = game.play(player1, player2)
print(f"\nThe relationship result between {player1} and {player2} is: ✨ {result} ✨")
```
