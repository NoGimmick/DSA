# Python Built-ins for Writing Pythonic Code

This notebook summarizes commonly used built-ins and language features that make Python code cleaner, more readable, and more modular.

## enumerate()

**Purpose:** Iterate with both index and value instead of range(len(...)).

## zip()

**Purpose:** Iterate over multiple iterables together.

## any()

**Purpose:** Returns True if at least one item is truthy.

## all()

**Purpose:** Returns True if every item is truthy.

## sorted()

**Purpose:** Returns a sorted copy without modifying the original.

## reversed()

**Purpose:** Reverse iteration without manual indexing.

## sum(), min(), max()

**Purpose:** Aggregate numeric collections.

## str.join()

**Purpose:** Efficiently concatenate strings.

## str.split()

**Purpose:** Split text into tokens.

## str.strip()

**Purpose:** Remove leading/trailing whitespace.

## dict.get()

**Purpose:** Retrieve values with a default.

## set()

**Purpose:** Store unique values and remove duplicates.

## collections.Counter

**Purpose:** Count frequencies.

## collections.defaultdict

**Purpose:** Avoid repetitive key initialization.

## pathlib.Path

**Purpose:** Modern filesystem paths.

## with

**Purpose:** Safely manage resources like files.

## List comprehensions

**Purpose:** Compact list creation.

## Generator expressions

**Purpose:** Lazy evaluation for functions like sum().

## isinstance()

**Purpose:** Type checking that supports inheritance.

## map() and filter()

**Purpose:** Functional helpers; know them, but prefer comprehensions for readability.

# Suggested Learning Order

1. Master loops and conditionals.
2. Learn `enumerate()` and `zip()`.
3. Learn `any()` and `all()`.
4. Practice comprehensions.
5. Learn dictionaries, `dict.get()`, `Counter`, and `defaultdict`.
6. Learn file handling with `with` and `pathlib`.
7. Learn regular expressions (`re`).

The goal is not shorter code for its own sake, but code that clearly expresses intent.


# Project Recommendation

## 1. Log Parser ⭐ (Recommended first)

A log parser is an excellent learning project, especially for someone interested in security engineering.

Suggested milestones:
- Read a log file.
- Parse each line.
- Count log levels.
- Extract timestamps and IP addresses.
- Detect repeated failures.
- Search/filter logs.
- Export summaries to CSV or JSON.

Skills you'll naturally learn:
- File handling
- String processing
- Regular expressions
- Dictionaries
- `Counter`
- `Path`
- Error handling
- Modular programming

## 2. Resume Parser

Build this after the log parser. It introduces additional complexity such as PDF parsing, OCR, and NLP, making it a great second parsing project.
